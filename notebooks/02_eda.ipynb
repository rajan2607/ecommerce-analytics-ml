```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/cleaned/cleaned_online_retail.csv")
df.head()
```

```python
df.info()

```

```python
total_revenue = df["Revenue"].sum()
total_revenue
```

```python
top_products = (
    df.groupby("Description")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_products
```

```python
df.describe()
```

```python
df.groupby("Country")["Revenue"].sum()
```

```python
df.groupby("Country")["Revenue"] \
  .sum() \
  .sort_values(ascending=False) \
  .head(10)
```

```python
top_countries = (
    df.groupby("Country")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_countries
```

```python
top_countries.plot(kind="bar", figsize=(10,5))
plt.title("Top 10 Countries by Revenue")
plt.xlabel("Country")
plt.ylabel("Revenue")
plt.show()
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
df["YearMonth"] = df["InvoiceDate"].dt.to_period("M")
```

```python
monthly_revenue = (
    df.groupby("YearMonth")["Revenue"]
      .sum()
)

monthly_revenue.head()
```

```python
monthly_revenue.plot(figsize=(12,5))
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

## Time-Based Insights

- Monthly revenue shows clear seasonal patterns.
- Revenue increases significantly during holiday months.
- Business experiences fluctuations rather than steady linear growth.
